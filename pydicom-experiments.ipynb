```python
# load-data
ds = pydicom.dcmread("data/95240362_0004.dcm")
```

## Inspect the header

```python
# inspect-key-fields
print("Modality:", ds.Modality)
print("Manufacturer:", ds.Manufacturer)
print("Model:", ds.ManufacturersModelName)
print("Frames:", ds.NumberOfFrames)
print("Rows x columns:", ds.Rows, "x", ds.Columns)
print("Frame time (ms):", ds.FrameTime)
```

```python
# inspect-file-meta
ds.file_meta
```

## Access pixel data

The pixel data is JPEG Baseline compressed, so decoding may require an additional package (for example Pillow or pylibjpeg).

```python
# decode-pixel-array
frames = ds.pixel_array
print(frames.shape, frames.dtype)
```

## Visualize frames

```python
# plot-frames
idx = [0, len(frames) // 2, len(frames) - 1]

fig, axes = plt.subplots(1, len(idx), figsize=(15, 4))
for ax, i in zip(axes, idx):
    ax.imshow(frames[i])
    ax.set_title(f"Frame {i}")
    ax.axis("off")
plt.tight_layout()
plt.show()
```
